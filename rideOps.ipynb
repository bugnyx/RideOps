```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("ola_driver_scaler.csv")
```

```python
print(df.shape)
print(df.head())
```

```python
print(df.info())
```

```python
df = df.drop(columns=['Unnamed: 0'])
df.columns = df.columns.str.replace(' ', '_').str.lower()
```

```python
df = df.rename(columns={
	'mmm-yy' : 'reporting_date',
	'dateofjoining' : 'joining_date',
	'lastworkingdate' : 'last_working_date',
	})

df['reporting_date'] = pd.to_datetime(df['reporting_date'], format='%m/%d/%y')
df['joining_date'] = pd.to_datetime(df['joining_date'], format='%d/%m/%y')
df['last_working_date'] = pd.to_datetime(df['last_working_date'], format='%d/%m/%y')
```

```python
df.head(10)
```

```python
print(df.isnull().sum())
```

```python
#forward/backward fill for age and gender columns
df['age'] = df.groupby('driver_id')['age'].ffill().bfill()
df['gender'] = df.groupby('driver_id')['gender'].ffill().bfill()

#fallback
df['age'] = df['age'].fillna(df['age'].median()).astype(int)
df['gender'] = df['gender'].fillna(df['gender'].mode()[0]).astype(int)
```

```python
#convert gender to int
df['gender'] = df['gender'].astype(int)
```

```python
print(df.isnull().sum())
```

```python
#sorting
df = df.sort_values(by=['driver_id', 'reporting_date']).reset_index(drop=True)

df.head()
```

```python
#churned?
df['is_churned'] = df['last_working_date'].notnull().astype(int)

print("Remaining Null Counts:")
print(df.isnull().sum())
```

```python
# Tenure in months as of each reporting date
df['tenure_months'] = ((df['reporting_date'].dt.year - df['joining_date'].dt.year) * 12 + (df['reporting_date'].dt.month - df['joining_date'].dt.month)) + 1
```

```python
df.head(10)
```

```python
#previous month metrics per driver
df['prev_income'] = df.groupby('driver_id')['income'].shift(1)
df['prev_bv'] = df.groupby('driver_id')['total_business_value'].shift(1)
df['prev_rating'] = df.groupby('driver_id')['quarterly_rating'].shift(1)

# Month-over-Month percentage changes
df['income_change_pct'] = np.where(
    (df['prev_income'].notnull()) & (df['prev_income'] > 0),
    ((df['income'] - df['prev_income']) / df['prev_income']) * 100,
    0
)

df['bv_change_pct'] = np.where(
    (df['prev_bv'].notnull()) & (df['prev_bv'] > 0),
    ((df['total_business_value'] - df['prev_bv']) / df['prev_bv']) * 100,
    0
)

df['rating_change'] = df['quarterly_rating'] - df['prev_rating'].fillna(df['quarterly_rating'])

df.head()
```

```python
# Health Score Calculation
# Component sub-scores normalized between 0 and 1
income_score = np.clip((df['income_change_pct'] + 50) / 100, 0, 1)
bv_score = np.clip((df['bv_change_pct'] + 50) / 100, 0, 1)
rating_score = (df['quarterly_rating'] - 1) / 3.0
rating_shift_score = np.clip((df['rating_change'] + 2) / 4.0, 0, 1)
tenure_score = np.clip(df['tenure_months'] / 24.0, 0, 1)

# Composite Health Score (0 - 100 scale)
df['health_score'] = np.round((
    0.25 * income_score +
    0.25 * bv_score +
    0.20 * rating_score +
    0.15 * rating_shift_score +
    0.15 * tenure_score
) * 100, 2)

# Categorization
conditions = [
    (df['health_score'] >= 70),
    (df['health_score'] >= 50) & (df['health_score'] < 70),
    (df['health_score'] >= 35) & (df['health_score'] < 50),
    (df['health_score'] < 35)
]
choices = ['Healthy', 'Stable', 'Watch List', 'At Risk']

df['health_category'] = np.select(conditions, choices, default='Watch List')

df[['driver_id', 'reporting_date', 'health_score', 'health_category']].head()
```

```python
# Splitting data into relational tables

dim_driver = df[[
    'driver_id', 'gender', 'age', 'education_level',
    'city', 'joining_date', 'joining_designation'
]].drop_duplicates(subset=['driver_id']).reset_index(drop=True)

fact_driver_monthly = df[[
    'driver_id', 'reporting_date', 'income', 'grade',
    'total_business_value', 'quarterly_rating', 'last_working_date',
    'is_churned', 'tenure_months', 'income_change_pct',
    'bv_change_pct', 'rating_change', 'health_score', 'health_category'
]].copy()
```

```python
dim_driver.to_csv('dim_driver.csv', index=False)
fact_driver_monthly.to_csv('fact_driver_monthly.csv', index=False)
df.to_csv('ola_driver_cleaned.csv', index=False)
```

```python
print(f"dim_driver shape: {dim_driver.shape}")
print(f"fact_driver_monthly shape: {fact_driver_monthly.shape}")
print("Files exported successfully!")
```
